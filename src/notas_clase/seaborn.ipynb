## Usando *Seaborn*

```python
# Cargar la biblioteca de seaborn, matplotlib y pandas
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Consultando el conjunto de datos disponibles
sns.get_dataset_names()
```

```python
# Cargando el conjunto de datos penguins
penguins = sns.load_dataset("penguins")

# Desplegando las columnas del dataframe
penguins.columns 
```

### Scatter-plots

```python
# Graficando la relación x=bill_length_mm, y=bill_depth_mm
sns.scatterplot(data= penguins,x="bill_length_mm", y="bill_depth_mm")

# Mostrando la gráfica
plt.show()
```

#### Para agregar opciones *estéticas* a la gráfica

```python
# Graficando la relación x=bill_length_mm, y=bill_depth_mmy color con base en species
sns.scatterplot(data=penguins,x="bill_length_mm", y="bill_depth_mm",hue="species")

# mostrando la gráfica
plt.show()
```

Color de los puntos de forma homogénea

Se pueden combinar varios atributos estéticos en una misma gráfica para representar múltiples dimensiones de los datos.
> Por ejemplo, podemos asignar:

- Color según la especie (hue)
- Tamaño según la longitud del pico (size)
- Forma según el sexo del pingüino (style)
- Color constante para todos los bordes de los puntos (edgecolor)

```python
# Grafica de puntos combinando atributos
sns.scatterplot(data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",           # color según especie
    size="flipper_length_mm", # tamaño según longitud del aleta
    style="sex",             # forma según sexo
    edgecolor="purple", # borde constante
    alpha=0.5 # La transparencia
)
plt.show()
```

### Histogramas

```python
# Histograma de la variable body_mass_g
sns.histplot(data=penguins,x="body_mass_g")
plt.show()

# Cambiando el color del contorno de la barra y agregando transparencia
# y el número de intervalos con el parámetro bins
sns.histplot(data=penguins,x="body_mass_g", edgecolor="black", alpha=0.3, bins= 300)
plt.show()
```

### Boxplots

```python
# boxplot de la variable body_mass_g, y sex
sns.boxplot(data=penguins,x= "sex",y="body_mass_g")
plt.show()

# Cambiando el color de las cajas y el de las líneas
sns.boxplot(data=penguins, x="body_mass_g", y="sex", color="orange", boxprops=dict(edgecolor="green", alpha=0.3))
plt.show()
```

### Presentación de los gráficos

Títulos, etiquetas y leyendas

> Se pueden agregar parámetros nativos de *matplotlib*


- plt.title("Texto") ---> Agrega un título en la parte superior de la gráfica.
- plt.xlabel("Texto") ---> Coloca una etiqueta en el eje X (horizontal).
- plt.ylabel("Texto") ---> Coloca una etiqueta en el eje Y (vertical).
- plt.grid(True) ---> Muestra una cuadrícula de fondo para facilitar la lectura.
- plt.legend() ---> Muestra la leyenda cuando hay más de una línea o categoría.


Además, para solicitar opciones, se puede usar *legend*

```python
# Solicitando ayuda de la función legend
help(plt.legend)
```

```python
# Crear la gráfica base
sns.scatterplot(data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",           # color según especie
    size="flipper_length_mm", # tamaño según longitud del aleta
    style="sex",             # forma según sexo
    edgecolor="purple"        # borde constante
)

# Agregar títulos y etiquetas
plt.title("Relación entre longitud y ancho del pico", fontsize=14, fontweight="bold",fontstyle = "italic", color ="yellow")
plt.xlabel("Longitud del pico (mm)", fontsize=12, color='yellow')
plt.ylabel("Ancho del pico (mm)", fontsize=12, color='yellow')

# Ajustar su posición posición y cambiar tamaño de letra
plt.legend(
    bbox_to_anchor=(1.02, 1),   # posición fuera del eje
    loc='upper left',           # punto de anclaje de la leyenda
    borderaxespad=0,
    fontsize = 9,
    title_fontsize=2,
    labelcolor = "yellow",
    facecolor = 'darkgrey'

)

# Cambiando el fondo de la gráfica y la figuraxs
plt.gca().set_facecolor("blue")    # fondo de los ejes
plt.gcf().set_facecolor("darkblue")   # fondo de la figura

# Agregar cuadrícula
plt.grid(True, linestyle="--", alpha=0.9, color='yellow')

plt.show()
plt.savefig("pinguinos_romero",       # nombre del archivo
            dpi=500,                     # resolución en puntos por pulgada
            bbox_inches="tight") 
```

```python

# Definir un diccionario con estilo de fuente
estilo_letra = {
    'fontsize': 14,
    'color': 'darkblue',
    'fontweight': 'bold',
    'fontstyle': 'italic'
}

# Crear la gráfica base
sns.scatterplot(data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",           # color según especie
    size="flipper_length_mm", # tamaño según longitud del aleta
    style="sex",             # forma según sexo
    edgecolor="purple"        # borde constante
)

# Agregar títulos y aplicar el estilo_letra
plt.title("Relación entre longitud y ancho del pico", fontdict=estilo_letra)
plt.xlabel("Longitud del pico (mm)", fontsize=12)
plt.ylabel("Ancho del pico (mm)", fontsize=12)

# Ajustar su posición posición y cambiar tamaño de letra
plt.legend(
    bbox_to_anchor=(1.02, 1),   # posición fuera del eje
    loc='upper left',           # punto de anclaje de la leyenda
    borderaxespad=0,
    fontsize = 8,
    title_fontsize=10,
    labelcolor = "grey"
)

# Cambiando el fondo de la gráfica y la figuraxs
plt.gca().set_facecolor("lavender")    # fondo de los ejes
plt.gcf().set_facecolor("mintcream")   # fondo de la figura

# Agregar cuadrícula
plt.grid(True, linestyle="--", alpha=0.3)

plt.show()
```

### Más formas de agregar títulos

En Seaborn, una forma práctica de personalizar los títulos, etiquetas y apariencia general de las gráficas es mediante el uso del objeto de tipo Axes que devuelve cada función de Seaborn.

A partir de este objeto (por ejemplo, g), es posible acceder tanto al eje (Axes) como a la figura (Figure) asociada, y utilizar métodos como set_title(), set_xlabel(), set_ylabel() o set_facecolor() para modificar elementos específicos del gráfico, así como suptitle() para agregar un título general a la figura. El siguiente ejemplo muestra cómo aplicar estas funciones para personalizar la gráfica de dispersión (scatterplot) que hemos venido trabajando:

```python
# Crear la gráfica y guardar el objeto Axes en 'g'
g = sns.scatterplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",            # color según especie
    size="flipper_length_mm", # tamaño según longitud de aleta
    style="sex",              # forma según sexo
    edgecolor="purple"
)

# Acceder al eje y modificar títulos de los ejes
g.set_title("Relación entre longitud y ancho del pico", fontsize=14, fontweight="bold", fontstyle="italic", color="darkblue")
g.set_xlabel("Longitud del pico (mm)", fontsize=12)
g.set_ylabel("Ancho del pico (mm)", fontsize=12)

# Cambiar color de fondo del gráfico
g.set_facecolor("black")
g.figure.set_facecolor("lightgreen")

# Acceder a la figura asociada y agregar un título general
g.figure.suptitle("Visualización de características morfológicas de los pingüinos", fontsize=16, fontweight="bold", color='yellow')

# Ajustar la leyenda
g.legend(
    bbox_to_anchor=(1.02, 1),
    loc='upper left',
    borderaxespad=0,
    fontsize=8,
    title_fontsize=10,
    labelcolor="grey"
)

plt.show()
```

### Subplots 
Otra cosa muy interesante que permite seaborn es crear subplots en una figura, es decir varios ejes o "axes" esto permite dibujar en un mismo espacio varios plots y a su vez tener control de la figura en si 

```python
#creamos el eje ax en la figura fig ob 
fig_ob, (ax_1,ax_2) = plt.subplots(1,2,figsize=(20,6)) 

sns.scatterplot(data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    ax=ax_1,
    hue="species",           # color según especie
    size="flipper_length_mm", # tamaño según longitud del aleta
    style="sex",             # forma según sexo
    edgecolor="purple"        # borde constante
)

ax_1.set_title("Relación entre longitud y ancho del pico")
ax_1.set_xlabel("Longitud del pico (mm)")
ax_1.set_ylabel("Ancho del pico (mm)")
ax_1.grid(True, linestyle="--", alpha=0.5)
ax_1.legend(
    bbox_to_anchor=(1.02, 1),
    loc='upper left',
    borderaxespad=0,
    fontsize=8,
    title_fontsize=10,
    labelcolor="grey"
)



sns.scatterplot(data=penguins,
    x='flipper_length_mm',
    y='body_mass_g',
    ax=ax_2,
    hue="species",           # color según especie
    size="flipper_length_mm", # tamaño según longitud del aleta
    style="sex",             # forma según sexo
    edgecolor="purple"        # borde constante
)

ax_2.set_title("Relación entre longitud y ancho del pico")
ax_2.set_xlabel("Longitud del pico (mm)")
ax_2.set_ylabel("Ancho del pico (mm)")
ax_2.grid(True, linestyle="--", alpha=0.5) 
ax_2.legend(
    bbox_to_anchor=(1.02, 1),
    loc='upper left',
    borderaxespad=0,
    fontsize=8,
    title_fontsize=10,
    labelcolor="grey"
)


plt.show
```

A si mismo se puede iterar sobre distintos plots que generes en una misma figura, lo que hace seaborn al tener la sintaxis plt.subplots(num1, num2) es un array de numpy de dos dimensiones donde num1 son las filas y num2 las columnas por lo que los plots guardados son el producto de ambos mismos que se pueden iterar con un `axes.flat()` para reducir la dimensionalidad y combinado con un `enumerate()` puede haber mas control dentro de un ciclo for.

```python
#en esta ocasion guardamos las 4 figuras en un arrar "axes" 
fig_ob, axes = plt.subplots(2,2,figsize=(35,13)) 


dic_fig={
    1:["flipper_length_mm","bill_length_mm" ],
    2:["bill_length_mm","bill_depth_mm"],
    3:['flipper_length_mm','body_mass_g'],
    4:["bill_depth_mm",'body_mass_g']
}

#enumearte regresa inice elementp 
for i, ax in enumerate(axes.flat, start=1): 
    sns.scatterplot(
        data=penguins,
        x=dic_fig[i][0],
        y=dic_fig[i][1],
        ax=ax,
        hue="species",           # color según especie
        size="flipper_length_mm", # tamaño según longitud del aleta
        style="sex",             # forma según sexo
        edgecolor="purple"        # borde constante
    ) 
    ax.set_title(f"Relación entre {dic_fig[i][0]} y {dic_fig[i][1]}")
    ax.set_xlabel(f"{dic_fig[i][0]}")
    ax.set_ylabel(f"{dic_fig[i][1]}")
    ax.grid(True, linestyle="--", alpha=0.5)
    ax.legend(
        bbox_to_anchor=(1.02, 1),
        loc='upper left',
        borderaxespad=0,
        fontsize=8,
        title_fontsize=10,
        labelcolor="grey"
    )

plt.show()
```
