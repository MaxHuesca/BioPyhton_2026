## Crear una tabla 
CREATE TABLE: Comando para crear una tabla.

table_name: Nombre de la nueva tabla. Debe ser descriptivo, singular y siguiendo la convención de Nombres. (sin simbolos especiales, sin espacios, etc).

columnX_name: Nombre de vada una de las columnas de la tabla.

datatype: Especifica el tipo dell dato que la columna almacenara:

    INT : para enteros 

    VARCHAR(n): Para cadana de caracteres de a lo mas n characteres

    DATE:  para fechas

    DECIMAL(p, s) para numeros con con p digitos de los cuales s son decimales.

    FLOAT(p, s):  para numeros con con p digitos de los cuales s son decimales. Tienen mayor precisión que los decimales

    DOUBLE(p,s):  para numeros con con p digitos de los cuales s son decimales. Tienen mayor precisión que los float

constraints: Restricciones o reglas aplicables a la columna. Como las siguientes:

    NOT NULL: Asegura que la columna no pueda tener valores en  NULL .

    AUTO_INCREMENT: Automaticamente  asigna un valor entero unico que se autoincrementa en la columna cuando hay una inserción. (comunmente es usado con llaves primarias).

    PRIMARY KEY: Designa la columna (o conjunto de columnas) como identificador unidco de aa rnglon de la tabla

    DEFAULT value: Asigna un valor por  default cuando ningua valor es explicitamente asignado durante una inserción.

    FOREIGN KEY :  Para definir llaves foraneas. 


    

```python
import mysql.connector
import Credenciales

mydb = mysql.connector.connect(
  host=Credenciales.host,
  user=Credenciales.usuario,
  password=Credenciales.contrasena,
  database=Credenciales.database
)

# La sentencia SQL para crear la tabla ExpDiff
# Se añade "IF NOT EXISTS" para evitar errores si la tabla ya existe
CREATE_TABLE_QUERY = """
CREATE TABLE IF NOT EXISTS ExpDiff (
    ExpDiff_id INT AUTO_INCREMENT PRIMARY KEY,
    gene_id VARCHAR(50) UNIQUE,
    baseMean DECIMAL(40, 30),
    log2FC DECIMAL(40, 30),
    lfcSE DECIMAL(40, 30),
    stats DECIMAL(40, 30),
    pvalue DOUBLE(40,30),
    padj DOUBLE(40,30)
);
"""

mycursor = mydb.cursor()

mycursor.execute(CREATE_TABLE_QUERY)
print("La tabla 'ExpDiff' ha sido creada exitosamente o ya existía.")

if mycursor:
     mycursor.close()
     if mydb and mydb.is_connected():
            mydb.close()
            print("Conexión a MySQL cerrada.")
```

### Insertar datos dentro de una tabla

Inserción de datos en una tabla
La sintaxis para insertar datos en una tabla es:

Incluyendo los nombres de las columnas (recomendado):

INSERT INTO nombre_de_la_tabla (columna2, columna3, columna4)
VALUES (valor2, valor3, valor4);
Omitimos el primer valor asumiendo que la primera columna es autoincrement o tiene un valor por default.

Si vamos a introducir valores en todas las columnas:

INSERT INTO nombre_de_la_tabla 
VALUES (valor1, valor2, valor3, valor4);
Nosotros haremos un insert por cada elemento en la tabla de Resultados de la expresión diferencial.

```python

conn = mysql.connector.connect(
  host=Credenciales.host,
  user=Credenciales.usuario,
  password=Credenciales.contrasena,
  database=Credenciales.database
)

cursor = conn.cursor()

# Ruta del archivo
archivo = '../results/pasilla_ejemplo/matrix/ExpDifRes_v2.txt'

contador=0
# Leer y procesar el archivo
with open(archivo, 'r') as file:
    file.readline() # Esto lee y descarta la primera línea
    for linea in file:
        valores = linea.strip().split(',')
        query = "INSERT INTO ExpDiff (gene_id,baseMean, log2FC, lfcSE, stats, pvalue, padj) VALUES (%s, %s,%s, %s, %s, %s, %s)"
        cursor.execute(query, valores) 
        contador= contador+cursor.rowcount

# Confirmar los cambios.
conn.commit()
# Imprimir el número de filas afectadas
print(contador, "registros insertados.")

cursor.close()
conn.close()
print ( "Done")
```

### Consultar datos de la DB 
Consulta de datos de una o mas tablas
Mostrar todas las columnas y filas:

SELECT * FROM nombre_tabla;

Mostrar columnas específicas:

SELECT columna1, columna2 FROM nombre_tabla;

Filtrar filas con WHERE:
Usa la cláusula WHERE para establecer condiciones.

SELECT * FROM nombre_tabla WHERE columna = 'valor';

SELECT * FROM nombre_tabla WHERE numero_columna > 100;

Uniendo mas de una condicion con conectores logicos:

 SELECT columna1, columna2 FROM nombre_tabla
 WHERE columna = 'valor'
 AND numero_columna > 100;
Ordenar los resultados con ORDER BY:

SELECT * FROM nombre_tabla ORDER BY columna ASC; (ascendente)

SELECT * FROM nombre_tabla ORDER BY columna DESC; (descendente)
